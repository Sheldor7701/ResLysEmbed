```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.metrics import classification_report, roc_curve, auc
from sklearn.metrics import matthews_corrcoef, average_precision_score


# Load predicted labels and probabilities
labels = np.load('predicted_values.npy')
probabilities = np.load('predicted_probabilities.npy')

# Load true labels from CSV (assumes a column named 'label')
df = pd.read_csv('../DbPtm/Test/dbptm_benchmark_filtered.csv')
true_labels = df['label'].values



# Evaluate
accuracy = accuracy_score(true_labels, labels)
precision = precision_score(true_labels, labels)
recall = recall_score(true_labels, labels)
f1 = f1_score(true_labels, labels)
mcc = matthews_corrcoef(true_labels, labels)
auc = roc_auc_score(true_labels, probabilities)
aupr = average_precision_score(true_labels, probabilities)
confusion = confusion_matrix(true_labels, labels)
specificity = confusion[0, 0] / (confusion[0, 0] + confusion[0, 1]) if (confusion[0, 0] + confusion[0, 1]) > 0 else 0

# Print results
print(f'Accuracy: {accuracy:.4f}')
print(f'Precision: {precision:.4f}')
print(f'Recall: {recall:.4f}')
print(f'Specificity: {specificity:.4f}')
print(f'F1 Score: {f1:.4f}')
print(f'Matthews Correlation Coefficient: {mcc:.4f}')
print(f'ROC AUC: {auc:.4f}')
print(f'Average Precision: {aupr:.4f}')
print('Confusion Matrix:')
print(confusion)
```
